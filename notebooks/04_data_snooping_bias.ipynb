# Notebook 04: Data Snooping Bias

## Objective

Investigate the impact of data snooping bias in quantitative research.

Data snooping bias occurs when a researcher repeatedly tests numerous hypotheses, parameter combinations, or trading rules on the same historical dataset and subsequently reports only the most favorable outcome.

Although the selected strategy may appear highly profitable, its apparent success may simply reflect chance rather than genuine predictive power.

In this notebook, a large number of SMA crossover strategies are evaluated on the same dataset. The objective is to illustrate how extensive experimentation increases the likelihood of discovering apparently exceptional strategies that may fail to generalize to unseen data.
## Methodology

The experiment follows these steps:

1. Download historical daily price data for SPY.
2. Generate a large collection of SMA crossover strategies using multiple combinations of fast and slow moving averages.
3. Evaluate each strategy using identical performance metrics.
4. Rank strategies according to historical performance.
5. Compare the best-performing strategy with the original 50/200 SMA baseline.

By examining the entire search process rather than focusing solely on the top result, the effects of data snooping become visible.

```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
import os

sys.path.append("..")
from src.data_loader import load_data
from src.metrics import total_return,sharpe_ratio,max_drawdown
```

```python
def run_sma_backtest(ticker, fast=50, slow=200):
    df = load_data(ticker)

    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df["Fast_50"] = df["Close"].rolling(fast).mean()
    df["Slow_200"] = df["Close"].rolling(slow).mean()

    df["Signal"] = 0
    df.loc[df["Fast_50"] > df["Slow_200"], "Signal"] = 1

    df["Returns"] = df["Close"].pct_change()

    df["Strategy_Returns"] = (
        df["Signal"].shift(1) * df["Returns"]
    )

    df["Equity"] = (
        1 + df["Strategy_Returns"].fillna(0)
    ).cumprod()

    return df
```

```python
fast_windows = [x for x in range(5, 105, 5)]
slow_windows = [y for y in range(50, 305, 5)]

print("Strategies tested:",
      len(fast_windows) * len(slow_windows))
```

```python
results = []

for fast in fast_windows:
    for slow in slow_windows:

        if fast >= slow:
            continue

        df = run_sma_backtest(
            "SPY",
            fast,
            slow
        )

        results.append({
            "Fast": fast,
            "Slow": slow,
            "Return": round(
                total_return(df["Equity"]) * 100,
                2
            ),
            "Sharpe": round(
                sharpe_ratio(
                    df["Strategy_Returns"].dropna()
                ),
                2
            )
        })
```

```python
results_df = pd.DataFrame(results)

results_df.sort_values(
    "Return",
    ascending=False
).head(10)
```

## Top Performing Strategies

After evaluating a large number of SMA combinations, several strategies appeared to deliver substantially stronger historical performance than the baseline strategy.

If only these top-performing configurations were reported, one might incorrectly conclude that exceptional predictive relationships had been discovered.

However, these results emerged only after extensive experimentation across many alternative hypotheses.

```python
baseline = results_df[
    (results_df["Fast"] == 50)
    &
    (results_df["Slow"] == 200)
]

best = results_df.loc[
    results_df["Return"].idxmax()
]

print("Baseline:")
print(baseline)

print("\nBest:")
print(best)
```

## Baseline versus Best Strategy

The traditional 50/200 SMA crossover was compared with the highest-performing strategy identified during the search process.

While the optimized strategy achieved superior historical performance, it was selected retrospectively after observing the outcomes of hundreds of competing alternatives.

As a result, the improvement may reflect statistical luck rather than a durable investment edge.

```python
print("Strategies tested:",
      len(results_df))

print("Best Return:",
      round(results_df["Return"].max(), 2),
      "%")

print("Median Return:",
      round(results_df["Return"].median(), 2),
      "%")
```

## Interpretation

This experiment demonstrates the central danger of data snooping bias.

As the number of tested hypotheses increases, the probability of finding apparently extraordinary results due purely to random variation also increases.

A researcher who reports only the strongest-performing strategy creates an illusion of predictive skill, while concealing the many unsuccessful experiments that preceded it.

The issue is not necessarily intentional misconduct. Rather, data snooping often arises naturally through curiosity-driven exploration and iterative experimentation.

Consequently, robust quantitative research requires transparency regarding the number of tests performed and independent validation using unseen data.
## Key Findings

- Hundreds of SMA parameter combinations were evaluated using the same historical dataset.
- Several strategies appeared to outperform the baseline strategy substantially.
- Selecting the single best-performing strategy after repeated experimentation introduces data snooping bias.
- Historical outperformance alone does not guarantee future predictive power.
- Out-of-sample testing and walk-forward validation are essential safeguards against false discoveries.

# Conclusion

Data snooping bias represents one of the most subtle yet pervasive threats in quantitative research.

The more opportunities researchers give themselves to search for profitable patterns, the easier it becomes to mistake noise for genuine signal.

The purpose of backtesting is not to discover the most impressive historical result. Rather, it is to identify strategies capable of surviving contact with future market conditions.

Therefore, disciplined validation procedures, skepticism toward exceptional results, and transparency regarding the research process remain fundamental principles of sound quantitative analysis.
